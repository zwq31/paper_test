```python
from datasets import load_from_disk
import os
import json

# 假设你保存的是 test 文件夹（可以改为你的路径）
hf_data_dir = "../data/live_bench/math/test"
output_dir = "../data/live_bench/math/web_of_lies_v2"
os.makedirs(output_dir, exist_ok=True)
output_jsonl = os.path.join(output_dir, "question.jsonl")

dataset = load_from_disk(hf_data_dir)

with open(output_jsonl, "w", encoding="utf-8") as f:
    for item in dataset:
        f.write(json.dumps({
            "question_id": item["question_id"],
            "turns": item["turns"]
        }, ensure_ascii=False) + "\n")

print(f"✅ 已保存为 LiveBench 可读的 JSONL: {output_jsonl}")
```
